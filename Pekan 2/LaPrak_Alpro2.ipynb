## 2.1 Dasar Teori

Python merupakan bahasa pemrograman yang menggunakan berbagai jenis tipe data untuk menyimpan dan mengolah informasi. Tipe data merupakan bagian penting dalam pemrograman karena menentukan jenis nilai yang dapat disimpan dan operasi yang dapat dilakukan terhadap nilai tersebut. Dalam Python, beberapa tipe data dasar yang sering digunakan adalah string, numerik, dan boolean.

String merupakan tipe data yang digunakan untuk menyimpan data berupa teks atau kumpulan karakter. String dapat ditulis menggunakan tanda petik tunggal (`' '`) maupun tanda petik ganda (`" "`). Contohnya adalah `"Python"` atau `"Praktikum"`.

Tipe data numerik digunakan untuk menyimpan nilai berupa angka. Python memiliki beberapa jenis data numerik, di antaranya integer (`int`) untuk bilangan bulat dan float (`float`) untuk bilangan desimal. Tipe data numerik dapat digunakan dalam berbagai operasi matematika, seperti penjumlahan, pengurangan, perkalian, dan pembagian.

Boolean merupakan tipe data yang hanya memiliki dua nilai, yaitu `True` dan `False`. Tipe data ini biasanya digunakan dalam kondisi atau perbandingan untuk menentukan apakah suatu pernyataan bernilai benar atau salah.

Selain tipe data, Python juga menggunakan variabel untuk menyimpan suatu nilai. Variabel dapat berisi berbagai jenis data dan nilainya dapat diubah selama program berjalan. Sementara itu, konstanta merupakan nilai yang secara konsep ditetapkan agar tidak berubah selama program digunakan. Python tidak memiliki sintaks khusus untuk membuat konstanta, sehingga konstanta biasanya ditulis menggunakan nama dengan huruf kapital sebagai penanda, misalnya `PI = 3.14`.

Python juga menyediakan fungsi `input()` yang digunakan untuk menerima masukan dari pengguna. Data yang diterima melalui `input()` secara default memiliki tipe string sehingga dapat dikonversi ke tipe data lain apabila diperlukan, misalnya menggunakan `int()` atau `float()`.

Dengan memahami string, numerik, boolean, variabel, konstanta, dan input, pengguna dapat memahami cara data direpresentasikan, disimpan, dan digunakan dalam program Python. Konsep-konsep tersebut menjadi dasar untuk membuat program yang lebih kompleks pada materi pemrograman selanjutnya.

### 2.2.2 Program Numerik "Numerik_2611533008"

```python
a_3008 = 100
print ("The type of variable having value", a_3008, "is", type(a_3008))

# variable float value
b_3008 = 10.2345
print ("the type of variable having value", b_3008, "is", type(b_3008))

# variable complex value
c_3008 = 100+3j
print ("the type of variable having value", c_3008, "is", type(c_3008))
```

**Penjelasan:**

3. Bagian ketiga digunakan untuk menampilkan hasil perhitungan menggunakan fungsi print().

### 2.2.3 Program Boolean "Boolean_2611533008"

```python
is_lulus_3008 = True
is_cumlaude_3008 = True

nilai_3008 = 85
batas_lulus_3008 = 75

status_kelulusan_3008 = nilai_3008 >= batas_lulus_3008 

print("==== Check Kelulusan ====")
print("nilai:", nilai_3008)
print("Apakah lulus:", status_kelulusan_3008)
if is_lulus_3008 and is_cumlaude_3008:
    print("Selamat, Anda lulus dengan predikat Cum Laude")
```

**Penjelasan:**

1. Bagian pertama program digunakan untuk mendefinisikan variabel dengan nilai boolean, yaitu True atau False.
2. Bagian kedua digunakan untuk melakukan proses atau perbandingan yang menghasilkan nilai boolean.
3. Bagian ketiga digunakan untuk menampilkan hasil boolean menggunakan fungsi print().

### 2.2.4 Program Input "Input_2611533008"

```python
panjang_3008 = input('masukkan nilai panjang: ')
lebar_3008 = input('masukkan nilai lebar: ')
luas_3008 = int(panjang_3008) * int(lebar_3008)
print("luas=", luas_3008)
```

**Penjelasan:**

1. Bagian pertama program digunakan untuk mendefinisikan variabel dan menerima data dari pengguna menggunakan fungsi input().
2. Bagian kedua digunakan untuk mengolah data yang telah dimasukkan oleh pengguna.
3. Bagian ketiga digunakan untuk menampilkan hasil pengolahan data menggunakan fungsi print().

### 2.2.5 Program Konstanta "Konstanta_2611533008"

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3008 = float(input('masukkan nilai jari-jari:'))
luas_3008 = PI * jari_3008 * jari_3008
print ("Luas lingkaran dengan jari-jari %.2f adalah %.2f" %(jari_3008, luas_3008))
```

**Penjelasan:**

1. Bagian pertama program digunakan untuk mendefinisikan nilai yang akan digunakan sebagai konstanta.
2. Bagian kedua digunakan untuk menggunakan nilai konstanta dalam proses atau perhitungan program.
3. Bagian ketiga digunakan untuk menampilkan hasil penggunaan konstanta menggunakan fungsi print().
