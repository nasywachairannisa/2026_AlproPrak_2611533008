## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub mempelajari beberapa konsep dasar dalam pemrograman, yaitu penggunaan bahasa Python, Visual Studio Code sebagai alat untuk menulis kode, serta GitHub sebagai platform untuk menyimpan dan mengelola kode program secara daring.

1. **Python**
   Python adalah salah satu bahasa pemrograman tingkat tinggi yang mempunyai struktur dan sintaks yang relatif sederhana sehingga mudah dipelajari, khususnya bagi pemula. Python dapat dimanfaatkan untuk berbagai keperluan, seperti membuat aplikasi, mengolah data, hingga mengembangkan perangkat lunak.

2. **Visual Studio Code**
   Visual Studio Code atau VS Code merupakan aplikasi code editor yang digunakan untuk membuat, menulis, dan mengubah kode program. VS Code mendukung berbagai jenis bahasa pemrograman, termasuk Python. Aplikasi ini juga dilengkapi dengan beberapa fitur pendukung, seperti terminal, debugging, dan integrasi dengan Git.

3. **Git**
   Git adalah sistem pengontrol versi atau *version control system* yang berfungsi untuk mencatat dan mengelola perubahan yang terjadi pada kode program. Penggunaan Git memungkinkan programmer untuk menyimpan beberapa versi program serta melihat riwayat perubahan yang telah dibuat.

4. **GitHub**
   GitHub adalah platform berbasis online yang digunakan untuk menyimpan dan mengelola *repository* Git. GitHub dapat dimanfaatkan sebagai tempat pencadangan kode, sehingga program dapat diakses melalui berbagai perangkat. Selain itu, GitHub juga mendukung kerja sama antara beberapa programmer dalam mengembangkan suatu proyek.

5. **Repository**
   Repository merupakan tempat yang digunakan untuk menyimpan suatu proyek beserta kode program dan file pendukung lainnya. Repository dapat berada di komputer secara lokal maupun disimpan secara online melalui GitHub.

6. **Commit dan Push**
   *Commit* merupakan proses untuk menyimpan catatan perubahan yang telah dilakukan pada repository lokal. Sementara itu, *push* merupakan proses mengirimkan perubahan dari repository lokal ke repository yang tersimpan di GitHub.
