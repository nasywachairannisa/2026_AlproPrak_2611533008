## 2.1 Dasar Teori

Struktur percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk menentukan tindakan program berdasarkan kondisi tertentu. Dalam Python, percabangan dapat digunakan agar program memberikan keluaran yang berbeda sesuai dengan nilai yang dimasukkan oleh pengguna. Pada praktikum ini, struktur yang dipelajari meliputi if, if-else, if-elif-else, multi-if, dan match-case.

Percabangan if digunakan untuk menjalankan perintah ketika suatu kondisi terpenuhi, sedangkan if-else digunakan untuk menangani dua kemungkinan kondisi. Sementara itu, if-elif-else digunakan ketika terdapat beberapa kondisi yang harus diperiksa secara berurutan. Pada program yang dibuat, struktur tersebut digunakan antara lain untuk menentukan kelayakan berdasarkan umur dan kepemilikan SIM serta menentukan kategori nilai berdasarkan rentang nilai.

Selain itu, praktikum mempelajari multi-if, yaitu beberapa kondisi if yang diperiksa secara terpisah. Struktur ini dapat digunakan ketika beberapa kondisi dapat terpenuhi secara bersamaan, seperti pada perhitungan diskon berdasarkan jumlah belanja, status member, dan kode promo. Praktikum juga menggunakan match-case untuk memilih keluaran berdasarkan nilai tertentu, seperti menentukan nama bulan berdasarkan angka yang dimasukkan pengguna.

Dengan memahami berbagai struktur percabangan tersebut, program dapat dibuat untuk memproses input dan menentukan keluaran berdasarkan kondisi yang telah ditetapkan. Penggunaan operator logika seperti and dan operator perbandingan seperti != juga membantu dalam menyusun kondisi yang lebih spesifik pada program.

## 2.2 Kode Program

### 2.2.1 Program if_elif_else1 "if_elif_else1_26115330008"




```python
umur_3008 = int(input("Input umur anda: "))
sim_3008 = input("Apakah anda sudah punya sim c:") [0]

if umur_3008 >= 17 and sim_3008 == 'y':
    print ("Anda sudah dewasa dan boleh bawa motor ")

elif umur_3008 >= 17 and sim_3008 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

elif umur_3008 < 17 and sim_3008 == 'y':
    print("anda belum cukup umur punya sim")
    
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")

print("Program selesai")
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan data umur dan status kepemilikan SIM C.
2. Bagian kedua digunakan untuk mengecek kondisi umur dan kepemilikan SIM C menggunakan percabangan if, elif, dan else.
3. Bagian ketiga digunakan untuk menentukan apakah pengguna sudah cukup umur dan diperbolehkan membawa motor berdasarkan kondisi yang diberikan.
4. Bagian keempat digunakan untuk menampilkan hasil pemeriksaan dan keterangan bahwa program telah selesai.

### 2.2.2 Program if 1 "if1_2611533008"

```python
ipk_3008 = float(input("Input ipk anda = "))

if ipk_3008 > 2.75 :
    print("Selamat anda lulus sangat memuaskan dengan ipk = " + str(ipk_3008)) 

print("program selesai")
```

4. Bagian keempat digunakan untuk menampilkan keterangan bahwa program telah selesai.

### 2.2.3 Program if 2"if2_2611533008"

```python
ipk_3008 = float(input("Input ipk anda = "))

if ipk_3008 > 2.75 :
    print("Selamat anda lulus sangat memuaskan dengan ipk = " + str(ipk_3008)) 

else: 
    print("Anda tidak lulus")
print("program selesai")
```

### 2.2.4 Program match_case "match_case1_2611533008"

```python
bulan_3008 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3008: 
    case 1:
        print ("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print(" angka tidak valid")
    
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan angka bulan dari 1 sampai 12 yang akan diperiksa.
2. Bagian kedua digunakan untuk mencocokkan angka bulan yang dimasukkan dengan pilihan `case` menggunakan percabangan `match-case`.
3. Bagian ketiga digunakan untuk menampilkan nama bulan sesuai dengan angka yang dimasukkan, seperti angka 1 menampilkan Januari dan angka 12 menampilkan Desember.
4. Bagian keempat digunakan untuk menampilkan keterangan bahwa angka tidak valid jika angka yang dimasukkan tidak sesuai dengan pilihan bulan 1 sampai 12 menggunakan `case _`.


### 2.2.5 Program multi_if 1 "multi_if1_2611533008"

```python
umur_3008 = int(input(" Masukkan umur anda: "))
sim_3008 = input("Apakah anda punya sim (y/t): ") [0]

if umur_3008 >= 17 and sim_3008 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_3008 >= 17 and sim_3008 != 'y':
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_3008 < 17 and sim_3008 == 'y' :
    print("Anda belum cukup umur punya sim")

if umur_3008 < 17 and sim_3008 != 'y':
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan data umur dan status kepemilikan SIM C.
2. Bagian kedua digunakan untuk mengecek kondisi umur dan kepemilikan SIM C menggunakan beberapa percabangan `if` yang berdiri sendiri.
3. Bagian ketiga digunakan untuk menentukan status pengguna berdasarkan kombinasi umur dan kepemilikan SIM C, yaitu sudah dewasa dan boleh membawa motor, sudah dewasa tetapi tidak boleh membawa motor, belum cukup umur tetapi memiliki SIM, atau belum cukup umur dan tidak memiliki SIM.
4. Bagian keempat digunakan untuk menampilkan hasil pemeriksaan sesuai dengan kondisi yang terpenuhi.


### 2.2.6 Program multi_if 2 "multi_if2_2611533008"

```python
# program menghitug diskon belanja

# input dari user
total_belanja_3008 = float(input("Masukkan total belanja (Rp): "))

# input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3008 = input("apakah anda member? (y/t): "). strip() .lower() 
is_member_3008 = input_member_3008 in ["y", "ya"]

# input status kode promo ( mengecek apakah user mengeti 'y atau 'ya')
input_promo_3008 = input("apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3008 = input_promo_3008 in [ "y", "ya"]

total_diskon_persen_3008 = 0

# multi -  IF terpisah : setiap kondisi diperiksa secara independen
# diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3008 > 1000000:
    total_diskon_persen_3008 += 10 # diskon belanja besar
    
if is_member_3008:
    total_diskon_persen_3008 += 5 # diskon member
    
if kode_promo_valid_3008:
    total_diskon_persen_3008 += 15 # diskon voucher
    
# menghitung nominal diskon dan total bayar
nominal_diskon_3008 = total_belanja_3008 * (total_diskon_persen_3008 / 100)
total_bayar_3008 =  total_belanja_3008 - nominal_diskon_3008

# output hasil

print("\n --- RINCIAN PEMBAYARAN ---")
print (f"Total diskon  : {total_diskon_persen_3008}% (Rp {nominal_diskon_3008:,.0f})")
print(f"Total bayar  : {total_bayar_3008:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3008}%")

# output: total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan data total belanja, status member, dan status kode promo.
2. Bagian kedua digunakan untuk memeriksa apakah pelanggan memenuhi syarat untuk mendapatkan diskon belanja besar, diskon member, dan diskon kode promo menggunakan beberapa `if` yang berdiri sendiri.
3. Bagian ketiga digunakan untuk menghitung total persentase diskon, nominal diskon, dan total pembayaran setelah diskon.
4. Bagian keempat digunakan untuk menampilkan rincian pembayaran berupa total persentase diskon, nominal diskon, dan total yang harus dibayar.


### 2.2.7 Program nilai "nilai_2611533008"

```python
nilai_3008 = int(input("Inputkan nilai angka= "))

if nilai_3008>= 81:
    print("A")

elif nilai_3008 >= 70:
    print("B")

elif nilai_3008 >= 60:
    print("C")

elif nilai_3008 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan nilai angka yang akan diperiksa.
2. Bagian kedua digunakan untuk mengecek nilai angka berdasarkan beberapa rentang menggunakan percabangan `if`, `elif`, dan `else`.
3. Bagian ketiga digunakan untuk menentukan nilai huruf berdasarkan nilai angka, yaitu A untuk nilai 81 ke atas, B untuk nilai 70–80, C untuk nilai 60–69, D untuk nilai 50–59, dan E untuk nilai di bawah 50.
4. Bagian keempat digunakan untuk menampilkan hasil nilai huruf sesuai dengan kondisi nilai angka yang dimasukkan.
