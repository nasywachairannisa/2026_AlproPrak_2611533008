## 2.1 Dasar Teori

Python merupakan salah satu bahasa pemrograman yang memiliki berbagai struktur dasar untuk mengatur jalannya suatu program. Salah satu struktur yang penting untuk dipahami adalah perulangan (looping). Perulangan digunakan untuk menjalankan suatu perintah secara berulang berdasarkan kondisi atau jumlah tertentu. Dengan adanya perulangan, penulisan program menjadi lebih sederhana karena programmer tidak perlu menuliskan perintah yang sama berkali-kali.

Salah satu jenis perulangan dalam Python adalah for. Perulangan for digunakan untuk melakukan iterasi terhadap suatu urutan nilai atau rentang tertentu. Pada program yang dikerjakan, for digunakan bersama fungsi range() untuk menentukan nilai awal, batas akhir, dan jumlah perulangan. Contohnya adalah for i in range(1, ulang_3008 + 1):, yang digunakan untuk melakukan perulangan dari angka 1 sampai batas yang telah dimasukkan oleh pengguna.

Fungsi range() digunakan untuk menghasilkan urutan bilangan yang dapat digunakan dalam proses perulangan. Penggunaan range() dapat disesuaikan dengan kebutuhan program, misalnya dimulai dari 0, dimulai dari 1, atau menggunakan batas tertentu. Hal tersebut terlihat pada program yang menggunakan range(ulang_3008) untuk menampilkan nomor perulangan dan range(1, ulang_3008 + 1) untuk menampilkan angka mulai dari 1.

Perulangan for juga dapat dikombinasikan dengan percabangan if untuk memberikan kondisi tertentu pada proses perulangan. Salah satu penerapannya adalah menentukan bilangan genap menggunakan operator modulus %. Jika hasil pembagian suatu bilangan dengan 2 adalah 0, maka bilangan tersebut termasuk bilangan genap dan dapat diproses lebih lanjut.

Selain perulangan biasa, terdapat konsep nested for, yaitu perulangan yang berada di dalam perulangan lainnya. Nested for dapat digunakan ketika sebuah proses membutuhkan perulangan pada baris dan kolom secara bersamaan. Pada program yang dikerjakan, konsep ini digunakan untuk membuat pola persegi menggunakan karakter *. Selain itu, nested for juga digunakan untuk membuat pola segitiga dengan mengatur jumlah spasi dan simbol pada setiap baris.

Dengan memahami konsep for, range(), percabangan, dan nested for, mahasiswa dapat menyusun program yang lebih terstruktur serta mampu menerapkan perulangan untuk menyelesaikan berbagai permasalahan pemrograman sederhana.

## 2.2 Kode Program

### 2.2.1 Program jumlah genap "jumlah_genap_26115330008"




```python
ulang_3008 = int(input("masukkan nilai batas: " ))

jumlah_3008 = 0
for i in range(1, ulang_3008 + 1):
    if i % 2 == 0:
        print(i, end="")
        jumlah_3008 = jumlah_3008 + i

        if i < ulang_3008:
            print("+", end="")

        else:
            print("=", jumlah_3008, end="")
print()
print("jumlah = ", jumlah_3008)
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan nilai batas perulangan dari pengguna dan mengubahnya menjadi tipe data integer. Variabel jumlah_3008 diinisialisasi dengan nilai 0 sebagai tempat menyimpan hasil penjumlahan.
2. Bagian kedua digunakan untuk melakukan perulangan menggunakan for dari angka 1 sampai nilai batas yang dimasukkan pengguna.
3. Bagian ketiga digunakan untuk mengecek bilangan genap menggunakan kondisi if i % 2 == 0. Jika angka tersebut genap, maka angka akan ditampilkan dan ditambahkan ke variabel jumlah_3008.
4. Bagian keempat digunakan untuk menampilkan tanda + di antara bilangan genap yang dijumlahkan. Jika sudah mencapai bilangan terakhir, program menampilkan tanda = dan hasil penjumlahannya.
5. Bagian kelima digunakan untuk menampilkan hasil akhir jumlah seluruh bilangan genap yang ditemukan dalam batas perulangan.

### 2.2.2 Program nested_for 1 "nested_for1_2611533008"

```python
batas_3008 = int(input("masukkan nilai batas: "))
for line in range (1, batas_3008 + 1):
    for j in range(1, (-1*line + batas_3008 + 1)):
        print(".", end="")
    print(line)
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan nilai batas dari pengguna. Nilai tersebut dikonversi menjadi tipe data integer dan disimpan dalam variabel batas_3008.
2. Bagian kedua digunakan untuk melakukan perulangan menggunakan for dari angka 1 sampai nilai batas_3008.
3. Bagian ketiga digunakan untuk melakukan perulangan kedua (nested for) untuk menentukan jumlah titik yang akan ditampilkan pada setiap baris. Jumlah titik tersebut semakin berkurang sesuai dengan nilai line.
4. Bagian keempat digunakan untuk menampilkan karakter "." secara berulang pada satu baris. Penggunaan end="" membuat titik-titik tetap berada pada baris yang sama.
5. Bagian kelima digunakan untuk menampilkan nilai line setelah titik-titik selesai dicetak. Setelah itu, print() membuat program berpindah ke baris berikutnya.

### 2.2.3 nested_for_2"nested_for2_2611533008"

```python
batas_3008 = int(input("Masukkan nilai batas: "))

for i in range(1, batas_3008 + 1):
    for j in range(1, batas_3008 + 1):
        print("*", end="")
    print() # pindah ke baris berikutnya
```

**Penjelasan:**
1. Bagian pertama digunakan untuk memasukkan nilai batas dari pengguna. Nilai tersebut dikonversi menjadi tipe data integer dan disimpan dalam variabel batas_3008.
2. Bagian kedua digunakan untuk melakukan perulangan pertama menggunakan for. Perulangan berjalan dari angka 1 sampai batas_3008 dan disimpan dalam variabel i.
3. Bagian ketiga digunakan untuk melakukan perulangan kedua (nested for). Perulangan ini berjalan dari 1 sampai batas_3008 dan menggunakan variabel j.
4. Bagian keempat digunakan untuk mencetak karakter "*" sebanyak batas_3008 kali. end="" digunakan agar setiap tanda bintang tetap tercetak pada baris yang sama.
5. Bagian kelima menggunakan print() untuk berpindah ke baris berikutnya setelah perulangan j selesai. Dengan adanya perulangan i dan j, program akan membentuk pola persegi dari karakter "*".

### 2.2.4 Program nested_for_3 "nested_for3_2611533008"

```python
batas_3008 = int(input("masukkan nilai batas:"))
for i in range(1, batas_3008 + 1):
    for j in range(batas_3008 + 1):
        print(i+j, end=" ")
    print() # pindah ke baris berikutnya
```

3. Bagian ketiga digunakan untuk melakukan perulangan kedua (nested for) menggunakan variabel j. Perulangan ini berjalan dari 0 sampai batas_3008.
4. Bagian keempat digunakan untuk menghitung dan menampilkan hasil penjumlahan i + j. Hasilnya ditampilkan pada baris yang sama karena menggunakan end=" ".
5. Bagian kelima menggunakan print() untuk berpindah ke baris berikutnya setelah perulangan j selesai. Dengan adanya dua perulangan, program akan menampilkan pola angka berdasarkan hasil penjumlahan nilai i dan j.


### 2.2.5 Program nested_for4 "nested_for4_2611533008"

```python
tinggi_3008 = int(input("masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3008 % 2 != 0:
    print("tinggi harus bilangan genap")
else:
    a_3008 = tinggi_3008
    c_3008 = a_3008
    lebar_3008 = (2 * tinggi_3008) - 2

    for i in range(1, tinggi_3008 + 1):
        b_3008 = c_3008 + 1

        for j in range(1, lebar_3008 + 1):

            # baris atas dan bawah
            if i == 1 or i == tinggi_3008:
                if j == 1 or j == lebar_3008:
                    print("#", end="")
                else:
                    print("=", end="")

            # baris isi
            else:
                if j == 1 or j == lebar_3008:
                    print("|", end="")
                else:
                    if j == c_3008:
                        print("<", end="")
                    elif j == b_3008:
                        print(">", end="")
                    elif j == (lebar_3008 - c_3008):
                        print("<", end="")
                    elif j == (lebar_3008 - c_3008 + 1):
                        print(">", end="")
                    elif j > b_3008 and j < (lebar_3008 - c_3008):
                        print(".",end="")
                    else:
                        print(" ", end="")

        print()

        # logika asli java
        a_3008 -= 2
        
        if a_3008 <= 0:
            c_3008 = (-a_3008) + 2
        else:
            c_3008 = a_3008
```

**Penjelasan:**

1. Bagian pertama digunakan untuk memasukkan tinggi pola dari pengguna dan mengecek apakah nilai tersebut merupakan bilangan genap. Jika nilai yang dimasukkan ganjil, program akan menampilkan pesan bahwa tinggi harus berupa bilangan genap.
2. Bagian kedua digunakan untuk melakukan inisialisasi variabel a_3008, c_3008, dan lebar_3008. Variabel tersebut digunakan untuk menentukan ukuran serta posisi pola yang akan dibuat.
3. Bagian ketiga digunakan untuk melakukan perulangan menggunakan for dan nested for. Perulangan pertama menentukan baris, sedangkan perulangan kedua menentukan posisi karakter pada setiap baris. Percabangan if, elif, dan else digunakan untuk menentukan karakter seperti #, =, |, <, >, ., dan spasi sehingga membentuk pola tertentu.
4. Bagian keempat digunakan untuk mengubah nilai a_3008 dan c_3008 setelah setiap baris selesai. Perubahan nilai tersebut berfungsi untuk mengatur posisi pola pada baris berikutnya sehingga pola dapat terbentuk sesuai dengan desain yang diinginkan.

### 2.2.8 Program perulangan_for3 "perulangan_for3_2611533008"

```python
ulang_3008 = int(input("Masukkan jumlah perulangan: "))

jumlah_3008 = 0
for i in range(1, ulang_3008+1):
    print(i,end="")
    jumlah_3008  = jumlah_3008 + i
    
    if i < ulang_3008:
        print("+", end="")
    else:
        print("=", end="")
    
print()
print("jumlah=", jumlah_3008)
```

### 2.2.9 Program latihan 5 "latihan5_2611533008"

```python
tinggi_3008 = int(input("Masukkan tinggi segitiga: "))

for i in range(1, tinggi_3008 + 1):
    for j in range(tinggi_3008 - i):
        print(" ", end="")
    
    for j in range(i):
        print("* ", end="")
    
    print()
```

4. Bagian keempat digunakan untuk mencetak simbol * sesuai dengan nomor baris. Semakin besar nilai i, semakin banyak simbol * yang ditampilkan. Setelah itu, print() digunakan untuk berpindah ke baris berikutnya sehingga terbentuk pola segitiga.

Jika total pembayaran tidak lebih dari Rp300.000, program menampilkan ucapan terima kasih telah berkunjung. Setelah seluruh proses selesai, program menampilkan tulisan “Program selesai” yang menunjukkan bahwa program telah selesai dijalankan.
